```python
def compute_daily_averages(filtered_data):
    return filtered_data.resample('H').median()

hyy_data = data_fetched_hyy

hyy_data = compute_daily_averages(hyy_data)
```

```python
hyy_data['HYY_META.CO2672'].head()
```

```python
hyy_data['HYY_META.CO2672'] = hyy_data['HYY_META.CO2672'].combine_first(hyy_data['HYY_META.CO2icos672'])
```

```python
hyy_winter = hyy_df[(hyy_df.index.month == 12) | (hyy_df.index.month <= 2)]
hyy_spring = hyy_df[(hyy_df.index.month >= 3) & (hyy_df.index.month <= 5)]
hyy_autumn = hyy_df[(hyy_df.index.month >= 9) & (hyy_df.index.month <= 11)]
hyy_summer = hyy_df[(hyy_df.index.month >= 6) & (hyy_df.index.month <= 8)]
```

```python
import matplotlib.pyplot as plt

# Visualization of trends
plt.figure(figsize=(12, 6))
plt.scatter(hyy_winter.index, hyy_winter['HYY_META.O3672'], label='Winter O3', color='blue', marker='o', s=50)
plt.scatter(hyy_winter.index, hyy_winter['HYY_META.O3672'], label='Winter O3', color='red', marker='*', s=50)
plt.scatter(hyy_winter.index, hyy_winter['HYY_META.O3672'], label='Winter O3', color='green', marker='x', s=50)
plt.scatter(hyy_winter.index, hyy_winter['HYY_META.O3672'], label='Winter O3', color='violet', marker='+', s=50)

plt.title('O3 in Different Season over Year')
plt.xlabel('Date')
plt.ylabel('O3')
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.tight_layout()
plt.yscale('log')
plt.show()
```

```python
def remove_outliers_iqr(df, column_name):
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df_filtered = df[(df[column_name] >= lower_bound) & (df[column_name] <= upper_bound)]
    return df_filtered
#var_dat = var_data.dropna()
var_df = remove_outliers_iqr(var_df, 'VAR_META.NOX_1')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Extract the CO2 data
var_CO2_df = var_df_d['VAR_EDDY.av_c'].dropna()
hyy_CO2_df = hyy_df_d['HYY_META.CO2672'].dropna()

# Define a function for polynomial fitting
def polynomial(x, a, b, c):
    return a * x**2 + b * x + c

# Fit polynomial models
x_var = np.arange(len(var_CO2_df))
x_hyy = np.arange(len(hyy_CO2_df))

poly_fit_var, covar_var = curve_fit(polynomial, x_var, var_CO2_df.values)
poly_fit_hyy, covar_hyy = curve_fit(polynomial, x_hyy, hyy_CO2_df.values)

# Calculate standard errors of the parameters
std_err_var = np.sqrt(np.diag(covar_var))
std_err_hyy = np.sqrt(np.diag(covar_hyy))

# Generate fitted values
poly_fit_var_values = polynomial(x_var, *poly_fit_var)
poly_fit_hyy_values = polynomial(x_hyy, *poly_fit_hyy)

# Calculate R-squared values
def calculate_r_squared(y_true, y_pred):
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    return 1 - (ss_res / ss_tot)

r_squared_var = calculate_r_squared(var_CO2_df.values, poly_fit_var_values)
r_squared_hyy = calculate_r_squared(hyy_CO2_df.values, poly_fit_hyy_values)

# Print the fitting parameters, their standard errors, and R-squared values
print("VAR Polynomial Fit Parameters, Standard Errors, and R-squared:")
for i, (param, err) in enumerate(zip(poly_fit_var, std_err_var)):
    print(f"Parameter {i}: {param:.4f} ± {err:.4f}")
print(f"R-squared: {r_squared_var:.4f}")

print("\nHYY Polynomial Fit Parameters, Standard Errors, and R-squared:")
for i, (param, err) in enumerate(zip(poly_fit_hyy, std_err_hyy)):
    print(f"Parameter {i}: {param:.4f} ± {err:.4f}")
print(f"R-squared: {r_squared_hyy:.4f}")

# Plot the results
fig, ax = plt.subplots(1, 1, figsize=(24, 12))

ax.plot(hyy_CO2_df.index, hyy_CO2_df.values, marker='o', linestyle='-', label='HYY CO2')
ax.plot(var_CO2_df.index, var_CO2_df.values, marker='o', linestyle='-', label='VAR CO2')

# Plot polynomial fits
ax.plot(var_CO2_df.index, poly_fit_var_values, linestyle=':', color='blue', label='VAR Polynomial Fit')
ax.plot(hyy_CO2_df.index, poly_fit_hyy_values, linestyle=':', color='orange', label='HYY Polynomial Fit')

# Add text annotations for the polynomial fit parameters, their standard errors, and R-squared values
var_text = '\n'.join([f'VAR Param {i}: {param:.4f} ± {err:.4f}' for i, (param, err) in enumerate(zip(poly_fit_var, std_err_var))])
var_text += f'\nR-squared: {r_squared_var:.4f}'
hyy_text = '\n'.join([f'HYY Param {i}: {param:.4f} ± {err:.4f}' for i, (param, err) in enumerate(zip(poly_fit_hyy, std_err_hyy))])
hyy_text += f'\nR-squared: {r_squared_hyy:.4f}'

# Position the labels in different places
ax.text(0.55, 0.98, var_text, transform=ax.transAxes, fontsize=12, verticalalignment='top', color='blue')
ax.text(0.75, 0.98, hyy_text, transform=ax.transAxes, fontsize=12, verticalalignment='top', color='orange')

ax.set_title('Line Plot of CO2 by Year with Polynomial Fits', fontsize=16)
ax.set_xlabel('Year', fontsize=16)
ax.set_ylabel('CO2 [ppm]', fontsize=16)
ax.tick_params(axis='x', rotation=45, labelsize=14)
ax.tick_params(axis='y', labelsize=14)
ax.grid(True)
ax.legend(loc='best')

plt.tight_layout()
plt.show()
```

```python
from utils.data_extract import fetch_combined_data

# Use the year and will start from starting to ending time
variables = ['HYY_META.T168', 'HYY_META.RH168', 'HYY_META.WSU168', 'HYY_META.WDU168']
start_year = 1996
end_year = 2023

met_fetched_hyy = fetch_combined_data(variables, start_year, end_year)
```

```python
def compute_daily_averages(filtered_data):
    return filtered_data.resample('H').median()

hyy_met = met_fetched_hyy

hyy_met = compute_daily_averages(hyy_met)
```

```python
# Create temperature bins
temperature_bins_spring = pd.cut(hyy_met_spring['HYY_META.T168'], bins=[-10, 0, 10, 20, 30], include_lowest=True)

median_co2_spring = hyy_met_spring.groupby(temperature_bins_spring)[['HYY_META.O3672', 'HYY_META.NOx672', 'HYY_META.SO2672', 'HYY_META.CO672']].median()

# Plot side by side
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6))

# Spring plot
median_co2_spring.plot(kind='bar', ax=axes[0])
axes[0].set_title('Spring: Median CO2 Concentrations by Temperature Bins')
axes[0].set_xlabel('Temperature Range (°C)')
axes[0].set_ylabel('Median CO2')
axes[0].grid(axis='y')
axes[0].set_yscale('log')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from prophet import Prophet
import matplotlib.pyplot as plt

# Prepare the data for Prophet: reset the index to get the datetime column and rename it to 'ds'
df_prophet = var_df_d.reset_index()
df_prophet.rename(columns={df_prophet.columns[0]: 'ds', 'VAR_EDDY.av_c': 'y'}, inplace=True)

# Initialize and fit the model
model = Prophet()
model.fit(df_prophet)

# Make a future dataframe for forecasting
future = model.make_future_dataframe(periods=20)  # Forecasting 30 days into the future
forecast = model.predict(future)

# Plot the forecast
fig = model.plot(forecast)
plt.title('Prophet Forecast')
plt.xlabel('Date')
plt.ylabel('Value')
plt.show()
```
